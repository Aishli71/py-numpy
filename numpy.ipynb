1. Conversion from Python structures
   You can create a NumPy array from existing Python data structures like lists or tuples.

```python
import numpy as np
a = np.array([1,2,3,4])
print(a)
```

```python
import numpy as np
a = np.array(("apple","orange","mango"))
print(a)
```
